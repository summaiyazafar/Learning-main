#### Develop a Library Inventory Management System (Python Assignment)

```python
# 1. Data
library = {
    "Python Basics": {"author": "John Doe", "quantity": 5},
    "Data Science 101": {"author": "Jane Smith", "quantity": 3},
    "The Alchemist": {"author": "Paulo Coelho", "quantity": 4},
    "To Kill a Mockingbird": {"author": "Harper Lee", "quantity": 2},
    "1984": {"author": "George Orwell", "quantity": 6},
    "The Great Gatsby": {"author": "F. Scott Fitzgerald", "quantity": 3},
    "Harry Potter": {"author": "J.K. Rowling", "quantity": 8},
    "The Hobbit": {"author": "J.R.R. Tolkien", "quantity": 5},
    "Peer-e-Kamil": {"author": "Umera Ahmed", "quantity": 7},
    "Jannat Kay Pattay": {"author": "Nimra Ahmed", "quantity": 4},
    "C++ Programming": {"author": "Bjarne Stroustrup", "quantity": 3},
    "Introduction to AI": {"author": "Peter Norvig", "quantity": 2},
    "The Silent Patient": {"author": "Alex Michaelides", "quantity": 5},
    "Atomic Habits": {"author": "James Clear", "quantity": 10},
    "Rich Dad Poor Dad": {"author": "Robert Kiyosaki", "quantity": 6},
    "The Lean Startup": {"author": "Eric Ries", "quantity": 3},
    "A Brief History of Time": {"author": "Stephen Hawking", "quantity": 4},
    "Sapiens": {"author": "Yuval Noah Harari", "quantity": 5},
    "The 5 AM Club": {"author": "Robin Sharma", "quantity": 7},
    "Think and Grow Rich": {"author": "Napoleon Hill", "quantity": 8}
}

# 2. Display the data
def display_books():
    print("\n--- Current Library Inventory ---")
    
    # Check the data using on the loop
    for title, info in library.items():
        print(f"Book: {title} | Author: {info['author']} | Stock: {info['quantity']}")

# 3. Call the Function
display_books()
```

```python
# 3. Nayi kitab add karne ya purani ko restock karne ka function
def add_book():
    print("\n--- Add / Restock a Book ---")
    title = input("Enter book title: ").strip()
    
    # Check karenge ke kitab pehle se dictionary mein hai ya nahi
    if title in library:
        print(f"'{title}' pehle se majood hai.")
        extra_qty = int(input("Kitni mazeed kitabein add karni hain? "))
        library[title]['quantity'] += extra_qty # Purani quantity mein plus kar diya
        print(f"Stock updated! New quantity: {library[title]['quantity']}")
    else:
        # Agar nayi kitab hai to author aur quantity dono poochenge
        author = input("Enter author name: ").strip()
        qty = int(input("Enter quantity: "))
        
        # Dictionary mein naya data insert karna
        library[title] = {"author": author, "quantity": qty}
        print(f"'{title}' successfully added to the library!")
```

```python
display_books()
display_books()
```

```python
# 4. Kitab udhaar (borrow) karne ka function
def borrow_book():
    print("\n--- Borrow a Book ---")
    title = input("Enter the title of the book you want to borrow: ").strip()
    
    # 1. Check karein ke kitab library mein hai ya nahi
    if title in library:
        # 2. Check karein ke kya stock majood hai
        if library[title]['quantity'] > 0:
            library[title]['quantity'] -= 1  # Stock mein se ek kam kar diya
            print(f"✔️ You have successfully borrowed '{title}'. Enjoy reading!")
        else:
            print("❌ Sorry, this book is currently out of stock!")
    else:
        print("❌ This book is not available in our library.")
```

```python
display_books()   # 1. Pehle stock dekhein (Farz karein 'Atomic Habits' ki quantity 10 hai)
borrow_book()     # 2. 'Atomic Habits' borrow karein (Input mein naam likhein)
display_books()
```

```python
# 5. Kitab wapis (return) karne ka function
def return_book():
    print("\n--- Return a Book ---")
    title = input("Enter the title of the book you are returning: ").strip()
    
    # Agar kitab hamari library ki hai, to quantity barha dein
    if title in library:
        library[title]['quantity'] += 1
        print(f"✔️ Thank you! '{title}' has been returned and stock updated.")
    else:
        # Agar koi aisi kitab de raha hai jo list mein nahi thi
        print("This book is not in our record. Adding it as a new book.")
        author = input("Enter author name: ").strip()
        library[title] = {"author": author, "quantity": 1}
        print(f"✔️ '{title}' added to the library.")
```

```python
return_book()
```

```python
# 6. Main Program Menu Loop
while True:
    print("\n===== LIBRARY MANAGEMENT SYSTEM =====")
    print("1. View All Books")
    print("2. Add or Restock a Book")
    print("3. Borrow a Book")
    print("4. Return a Book")
    print("5. Exit")
    print("=====================================")
    
    choice = input("Select an option (1-5): ").strip()
    
    if choice == '1':
        display_books()
    elif choice == '2':
        add_book()
    elif choice == '3':
        borrow_book()
    elif choice == '4':
        return_book()
    elif choice == '5':
        print("\nThank you for using the Library System. Goodbye!")
        break  # Loop ko rokne aur program band karne ke liye
    else:
        print("❌ Invalid option! Please enter a number between 1 and 5.")
```
