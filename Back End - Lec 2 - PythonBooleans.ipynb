````python
# ============================================================
# 1. WHAT ARE PYTHON BOOLEANS?
# ============================================================
'''
Booleans are one of Python's simplest data types, but they appear everywhere: conditions, loops, comparisons, filtering, validation, and program logic.
A boolean value is either:
True
False
Notice the capitalization. `true` and `false` are invalid in Python.
'''
````

````python
# ============================================================
# 2. Creating Boolean Values
# ============================================================
is_raining = True
is_sunny = False

print(is_raining)
print(is_sunny)
'''
Check the type:
'''
print(type(True))
````

````python
# ============================================================
# 3. Booleans Are a Data Type
# ============================================================
a = True
b = False
print(type(a))
print(type(b))
````

````python
# ============================================================
# 4. Comparison Operators Return Booleans
# ============================================================
'''
Most booleans come from comparisons.
'''
# Equal
print(5 == 5)
print(5 == 10)

# Not Equal
print(5 != 10)
print(5 != 5)

# Greater Than
print(10 > 5)
print(5 > 10)

# Less Than
print(3 < 7)

# Greater Than or Equal
print(10 >= 10)
print(10 >= 15)

# Less Than or Equal
print(3 <= 3)
print(4 <= 2)
````

````python
# ============================================================
# 5. Boolean Variables in Conditions
# ============================================================
logged_in = True
if logged_in:
    print("Welcome!")

#Equivalent:
if logged_in == True:
    print("Welcome!")
#But the first version is preferred.
````

````python
# ============================================================
# 6. The `not` Operator
# ============================================================
'''
`not` reverses a boolean.
'''
print(not True)
print(not False)

#Example:
logged_in = False
if not logged_in:
    print("Please log in.")
````

````python
# ============================================================
# 7. The `and` Operator
# ============================================================
'''
Returns `True` only if both sides are true.
'''
print(True and True)
print(True and False)
print(False and True)
print(False and False)

#Example:
age = 25
has_ticket = True
if age >= 18 and has_ticket:
    print("Entry allowed")
````

````python
# ============================================================
# 8. The `or` Operator
# ============================================================
'''
Returns `True` if either side is true.
'''
print(True or True)
print(True or False)
print(False or True)
print(False or False)

#Example
is_admin = False
is_owner = True

if is_admin or is_owner:
    print("Access granted")
````

````python
# ============================================================
# 9. Combining `and`, `or`, and `not`
# ============================================================
age = 20
banned = False
if age >= 18 and not banned:
    print("Allowed")

#More complex:
premium = True
admin = False
if premium or admin:
    print("Special features enabled")
````

````python
# ============================================================
# 10. Order of Boolean Operations
# ============================================================
'''
Python evaluates:
1. `not`
2. `and`
3. `or`
'''

#Example:
result = True or False and False
print(result)

#Use parentheses when uncertain:
result = (True or False) and False
print(result)
````

````python
# ============================================================
# 11. Truthy and Falsy Values
# ============================================================
'''
Not everything must literally be `True` or `False`.
Python treats many values as truthy or falsy.
'''
# Falsy Values
'''
False
None
0
0.0
''
""
[]
{}
set()
()
'''

#Example1:
if 0:
    print("Runs")
else:
    print("Does not run")

#Example2:
if "":
    print("Runs")
else:
    print("Does not run")

# Truthy Values
'''
Almost everything else.
'''
#Example1:
if 100:
    print("Truthy")
#Example2:
if "hello":
    print("Truthy")
#Example3:
if [1, 2, 3]:
    print("Truthy")
````

````python
# ============================================================
# 12. Using `bool()`
# ============================================================
'''
Convert values to booleans.
'''
print(bool(1))
print(bool(0))
print(bool("hello"))
print(bool(""))
print(bool([1]))
print(bool([]))
````

````python
# ============================================================
# 13. Booleans and Numbers
# ============================================================
#In Python:
True == 1
False == 0
print(True + True)
print(True * 10)
print(False * 10)
print(int(True))
print(int(False))
````

````python
# ============================================================
# 14. Identity vs Equality
# ============================================================
#Equality:
print(True == 1)

#Identity:
print(True is 1)

#True == 1      # same value
#True is 1      # not same object/type

'''
if value is True:
only when you specifically need the boolean object itself.
if value:
is better.
'''
````

````python
# ============================================================
# 15. Chained Comparisons
# ============================================================
'''
Python supports chaining.
'''
x = 5
print(1 < x < 10)

#Equivalent to:
print(1 < x and x < 10)

#More examples:
age = 25
print(18 <= age < 65)
````

````python
# ============================================================
# 16. Membership Tests Return Booleans
# ============================================================

names = ["Alice", "Bob", "Charlie"]
print("Bob" in names)
print("David" in names)

#Negation:
print("David" not in names)
````

````python
# ============================================================
# 17. Boolean Functions
# ============================================================
'''
Functions can return booleans.
'''
def is_even(number):
    return number % 2 == 0
print(is_even(4))
print(is_even(5))

#Another example:
def is_adult(age):
    return age >= 18
````

````python
# ============================================================
# 18. Boolean Expressions in Loops
# ============================================================
count = 0
while count < 3:
    print(count)
    count += 1
#returns a boolean every iteration.
````

````python
# ============================================================
# 19. Short-Circuit Evaluation
# ============================================================
'''
Python sometimes stops early.
'''
# `and`
False and print("hello")
'''
`print()` never executes.
'''
#Example:
x = 0
if x != 0 and 10 / x > 1:
    print("Runs")
'''
Safe because:
x != 0
is false, so Python never evaluates:
10 / x
'''
# `or`
True or print("hello")
'''
Python already knows the answer is `True`.
The second expression is skipped.
'''
````

````python
# ============================================================
# 20. `and` and `or` Return Values
# ============================================================

#A common surprise:
print(5 and 10)
print(0 and 10)
print("" or "default")

# `and` returns the first falsy value or last value.
# `or` returns the first truthy value or last value.

username = input("Name: ") or "Guest"
'''
If user enters nothing:
username = "Guest"
'''
````

````python
# ============================================================
# 21. Custom Truthiness with Classes
# ============================================================

#You can define truthiness yourself.
class Wallet:
    def __init__(self, money):
        self.money = money

    def __bool__(self):
        return self.money > 0

#Usage:
wallet = Wallet(100)
if wallet:
    print("Has money")
````

````python
# ============================================================
# 22. Using `__len__`
# ============================================================

#If `__bool__` doesn't exist, Python may use `__len__`.
class Bag:
    def __init__(self, items):
        self.items = items

    def __len__(self):
        return len(self.items)

#Usage:
bag = Bag([])
if bag:
    print("Not empty")
else:
    print("Empty")
````
