* Read and display image.
* Image filtering
* Edge Detection
* Image Transformation

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

# Load the image
image = cv2.imread('flower.jfif')

if image is None:
   print('Error, Image not found.')
   exit()

cv2_imshow(image)
cv2.waitKey(0)
```

```python
# Convert to gray scale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
cv2.waitKey(0)
```

```python
# Image filtering (Noise Reduction)
# Gaussain Blur
blurred = cv2.GaussianBlur(gray, (5, 5), 0)
cv2_imshow(blurred)
cv2.waitKey(0)
```

```python
from google.colab.patches import cv2_imshow

# Median Filtering
median = cv2.medianBlur(gray, 5)

print("Median Filter")
cv2_imshow(median)
```

```python
# Edge Detection
edges= cv2.Canny(blurred, 100, 200)
cv2_imshow(edges)
```

```python
# Image Transformation
# Resize
resized = cv2.resize(image, (300, 300))
cv2_imshow(resized)
```

```python
# Rotation
height, width = image.shape[:2]
center = (width/2, height/2)
rotation_matrix = cv2.getRotationMatrix2D(center, 45, 1)
rotated = cv2.warpAffine(image, rotation_matrix, (width, height))
cv2_imshow(rotated)
```

```python
# Flip the image
flipped = cv2.flip(image, 1)
cv2_imshow(flipped)
```

```python
# Save the image
cv2.imwrite('flower_gray.jpg', gray)
cv2.imwrite('flower_blurred.jpg', blurred)
cv2.imwrite('flower_median.jpg', median)
```
