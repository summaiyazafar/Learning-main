## Python pandas library using for Web_Scraping 

```python
# step 1: Import the required libraries
import pandas as pd
import requests
from io import StringIO

# step 2: Load the URL for world population on wikipedia
url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

# step 3: Read the HTML tables from the URL
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)
tables = pd.read_html(StringIO(response.text))

# step 4: Check the number of tables found
print(f"Number of tables found: {len(tables)}")

# save this into csv
tables[0].to_csv("world_population_16072025.csv", index=False)

# step 5: Display the first few rows of the first table
tables[0].head()
```

```python
# Step 1: Import required libraries
import pandas as pd
import requests

# Step 2: Define the URL
url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

# Step 3: Add User-Agent header
headers = {
    "User-Agent": "Mozilla/5.0"
}

# Step 4: Send GET request
response = requests.get(url, headers=headers)

# Step 5: Check if request was successful
if response.status_code == 200:
    print("Website accessed successfully!")
else:
    print(f"Error: {response.status_code}")

# Step 6: Read all HTML tables from the webpage
tables = pd.read_html(response.text)

# Step 7: Display number of tables found
print(f"\nNumber of tables found: {len(tables)}")

# Step 8: Select the first table
df = tables[0]

# Step 9: Display first 5 rows
print("\nFirst 5 Rows:")
print(df.head())

# Step 10: Display DataFrame information
print("\nData Information:")
print(df.info())

# Step 11: Display column names
print("\nColumn Names:")
print(df.columns)

# Step 12: Save table to CSV file
df.to_csv("countries_population.csv", index=False)

print("\nData saved successfully as 'countries_population.csv'")
```
