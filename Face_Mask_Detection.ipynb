```python
catogies= ['with_mask', 'without_mask']
```

```python
print(catogies)
print(type(catogies))
```

```python
data= []
for category in catogies:
  path = os.path.join('/content/COVID_19_Dataset', category)

  label=catogies.index(category)

  for file in os.listdir(path):
    image_path = os.path.join(path, file)
    img = cv2.imread(image_path)
    img = cv2.resize(img, (224, 224))
    # cv2.imwrite(image_path, img)
    # print(img.shape)
    data.append([img, label])
```

```python
len(data)
```

```python
import random
random.shuffle(data)
```

```python
X= []
y= []

for features, label in data:
  X.append(features)
  y.append(label)

len(X)
```

```python
len(y)
```

```python
import numpy as np
X= np.array(X)
y= np.array(y)

X.shape
```

```python
y.shape
```

```python
X= X/255
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test= train_test_split(X, y, test_size= 0.2)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from keras.applications.vgg16 import VGG16
vgg= VGG16()
```

```python
vgg.summary()
```

```python
from keras import Sequential
model= Sequential()

for layer in vgg.layers[:-1]: # vgg is a seq model
  model.add(layer)

model.summary()
```

```python
for layer in model.layers:
  layer.trainable= False


model.summary()
```

```python
from keras.layers import Dense
model.add(Dense(1, activation= 'sigmoid'))
model.summary()
```

```python
model.compile(optimizer= 'Adam', loss= 'binary_crossentropy', metrics= ['accuracy'])
model.fit(X_train, y_train, epochs= 5, validation_data= (X_test, y_test))
```

```python
def detect_face_mask(img):
    # Normalize the image to 0-1 range, as the model was trained with normalized data
    img_normalized = img / 255.0

    # Use model.predict and correct the reshape dimensions to 224x224
    y_pred_proba = model.predict(img_normalized.reshape(1, 224, 224, 3))

    # Apply a threshold to get binary class prediction (0 or 1)
    y_pred = (y_pred_proba > 0.5).astype(int)

    return y_pred[0][0]
```
