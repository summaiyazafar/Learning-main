```python
#Here are the main ways you can use the `print()` statement/function in Python, with examples.

# Basic Print
print("Hello World")
```

```python
# Printing Multiple Values

print("Name:", "Ali", "Age:", 25)
```

```python
# Using Variables

name = "Henry"
age = 22

print(name)
print(age)
```

```python
# Using `sep` (separator)
#Controls what appears between values.

print("2026", "05", "23", sep="-")
```

```python
#Another example:

print("Python", "Java", "C++", sep=" | ")
```

```python
# Using `end`
#Controls what appears at the end of the print.

print("Hello", end=" ")
print("World")

#Without `end`, Python normally adds a newline.
```

```python
# Printing Without Newline

print("Loading...", end="")
```

```python
# Formatted Strings (f-strings)

name = "Ahmed"
marks = 95

print(f"{name} scored {marks} marks")
```

```python
# Using `.format()`

name = "Ayesha"
age = 21

print("My name is {} and I am {} years old".format(name, age))
```

```python
# Using `%` Formatting (Old Style)

name = "Hamza"
score = 88

print("Name: %s, Score: %d" % (name, score))
```

```python
# Printing Expressions

print(5 + 3)
print(10 * 2)
```

```python
# Printing Data Structures
## List

print([1, 2, 3])
```

```python
## Tuple

print((10, 20))
```

```python
## Dictionary

print({"name": "Ali", "age": 20})
```

```python
## Set

print({1, 2, 3})
```

```python
# Printing Objects

class Student:
    def __str__(self):
        return "Student Object"

s = Student()
print(s)
```

```python
# Pretty Printing
#Using the built-in `pprint` module.

from pprint import pprint

data = {
    "name": "Ali",
    "subjects": ["Math", "Physics", "CS"]
}

pprint(data)
```

```python
# Printing in Loops

for i in range(5):
    print(i)
```

```python
# Printing with Unpacking (`*`)

numbers = [1, 2, 3, 4]
print(*numbers)
```

```python
numbers = [1, 2, 3, 4]
print(numbers)
```

```python
#Custom separator:

print(*numbers, sep="-")
```

```python
# Redirecting Output

import sys

print("Error message", file=sys.stderr)
```

```python
# Combining Everything

name = "Ali"
age = 20
marks = [80, 90, 95]

print(f"Name: {name}", age, *marks, sep=" | ", end=" END\n")
```

```python
'''Internally, `print()` Syntax
print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)

Parameters:

* `objects` → items to print
* `sep` → separator between items
* `end` → what to print at the end
* `file` → where output goes
* `flush` → force immediate output'''
```

```python
# Basic f-string

name = "Ali"
print(f"Hello {name}")
```

```python
# Multiple Variables

name = "Sara"
age = 22

print(f"My name is {name} and I am {age} years old")
```

```python
# Expressions Inside f-string

a = 10
b = 5

print(f"Sum = {a + b}")
print(f"Product = {a * b}")
```

```python
# Float Formatting

pi = 3.14159265

print(f"Pi value: {pi:.2f}")
```

```python
# Padding and Alignment

name = "Ali"

print(f"{name:<10} left")
print(f"{name:>10} right")
print(f"{name:^10} center")
```

```python
# Percentage Formatting

score = 0.875

print(f"Percentage: {score:.2%}")
```

```python
# Using Functions in f-strings

name = "python"

print(f"{name.upper()}")
```

```python
# Debugging Style (Python 3.8+)

x = 5
y = 10

print(f"{x=}, {y=}")
```

```python
# Date Formatting

from datetime import datetime

today = datetime.now()

print(f"{today:%d-%m-%Y}")
```

```python
# Dictionary Values

student = {"name": "Ahmed", "marks": 95}

print(f"{student['name']} scored {student['marks']}")
```

```python
# Loop with f-strings

for i in range(3):
    print(f"Iteration {i}")
```

```python
# Multi-line f-string

name = "Ali"
age = 20

print(f"""
Name: {name}
Age: {age}
""")
```

```python
# Combining `sep`, `end`, and f-strings

name = "Sara"
score = 95

print(f"Student: {name}", f"Score: {score}", sep=" | ", end=" END\n")
```

```python
# Nested Formatting

value = 12.34567

print(f"Formatted value: {value:10.2f}")
```

```python
# Conditional Expression Inside f-string

age = 16

print(f"{'Adult' if age >= 18 else 'Minor'}")
```

```python
# All-in-One Example

name = "Ali"
age = 20
marks = 92.4567

print(f"Name: {name} | Age: {age} | Marks: {marks:.2f}")
```
