# 🔴 BBC News Real-Time Scraper
## Get Real Live Data - NOT Fake Data!

This scraper fetches **ACTUAL** BBC News headlines from the live website for today (2026-07-16)

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime
import json

print("✅ Libraries imported successfully!")
print(f"⏰ Current timestamp: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```

## Method 1: Live BBC News Scraper (Reliable HTML Parsing)

Scrapes real BBC News headlines from the live website

```python
print("🌐 SCRAPING REAL BBC NEWS DATA FROM LIVE WEBSITE")
print("=" * 80)

# Set up headers to identify as a real browser
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36',
    'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8',
    'Accept-Language': 'en-US,en;q=0.5',
    'Referer': 'https://www.google.com/'
}

bbc_news_url = 'https://www.bbc.com/news'

try:
    print(f"\n📡 Fetching: {bbc_news_url}")
    response = requests.get(bbc_news_url, headers=headers, timeout=15)
    response.raise_for_status()
    
    print(f"✅ Connection successful!")
    print(f"   Status Code: {response.status_code}")
    print(f"   Content Size: {len(response.content)} bytes")
    print(f"   Encoding: {response.encoding}")
    
except requests.exceptions.RequestException as e:
    print(f"❌ Error connecting to BBC News: {e}")
    exit()
```

```python
# Parse the HTML content
print("\n🔍 Parsing HTML content...")
soup = BeautifulSoup(response.content, 'html.parser')
print("✅ HTML parsed successfully")

# Extract real BBC News articles
real_articles = []
seen_headlines = set()

print("\n📰 Extracting articles from BBC News homepage...")
print("-" * 80)

# Strategy 1: Find article links by looking for specific BBC news patterns
all_links = soup.find_all('a', href=True)
print(f"Found {len(all_links)} total links on page")

article_count = 0
for link in all_links:
    if article_count >= 20:  # Get up to 20 real articles
        break
    
    try:
        href = link.get('href', '')
        
        # Filter for actual news article links
        if not href or href.startswith('#') or '/news/' not in href:
            continue
        
        # Get headline text
        headline = link.get_text(strip=True)
        
        # Skip if too short or duplicate
        if not headline or len(headline) < 8 or headline in seen_headlines:
            continue
        
        # Skip navigation items
        if any(skip in headline.lower() for skip in ['menu', 'home', 'sign in', 'search', 'more']):
            continue
        
        seen_headlines.add(headline)
        
        # Get full URL
        if not href.startswith('http'):
            full_url = 'https://www.bbc.com/news' + href
        else:
            full_url = href
        
        # Try to find description from nearby elements
        description = "N/A"
        parent = link.parent
        
        # Look in parent and siblings for description text
        if parent:
            for elem in parent.find_all(['p', 'span', 'div']):
                text = elem.get_text(strip=True)
                if (text and len(text) > 20 and len(text) < 200 and 
                    text != headline and 'cookie' not in text.lower()):
                    description = text[:150]
                    break
        
        # Add to articles list
        real_articles.append({
            'Headline': headline,
            'Description': description,
            'URL': full_url,
            'Scraped_Date': datetime.now().strftime('%Y-%m-%d'),
            'Scraped_Time': datetime.now().strftime('%H:%M:%S'),
            'Source': 'BBC News - Live Scrape'
        })
        
        article_count += 1
        print(f"✓ Article {article_count}: {headline[:60]}...")
    
    except Exception as e:
        continue

print("\n" + "=" * 80)
print(f"✅ Successfully extracted {len(real_articles)} REAL articles from BBC News!")
print("=" * 80)
```

## Display Real Data as DataFrame

```python
# Create DataFrame from real scraped data
df_real = pd.DataFrame(real_articles)

print("\n📊 BBC NEWS DATA - TODAY (REAL DATA)")
print("=" * 100)
print(f"\n📅 Date: {datetime.now().strftime('%Y-%m-%d')}")
print(f"⏰ Time: {datetime.now().strftime('%H:%M:%S')}")
print(f"📈 Total Articles: {len(df_real)}")
print(f"📝 Articles with Descriptions: {(df_real['Description'] != 'N/A').sum()}")
print("\n" + "=" * 100 + "\n")

# Display all articles
for idx, row in df_real.iterrows():
    print(f"\n🔹 Article #{idx + 1}")
    print(f"   📌 Headline: {row['Headline']}")
    print(f"   📝 Description: {row['Description']}")
    print(f"   🔗 URL: {row['URL']}")
    print("-" * 100)

print("\n" + "=" * 100)
print("COMPLETE DATA TABLE:")
print("=" * 100 + "\n")
print(df_real.to_string(index=False))
print("\n" + "=" * 100)
```

## Save Real Data to CSV

```python
# Save to CSV with timestamp
timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
csv_filename = f'BBC_News_Real_Data_{timestamp}.csv'

df_real.to_csv(csv_filename, index=False, encoding='utf-8')

print(f"\n💾 DATA SAVED TO CSV FILE")
print("=" * 100)
print(f"✅ File: {csv_filename}")
print(f"✅ Location: d:/archive/Plants/web_scraping/")
print(f"✅ Records: {len(df_real)}")
print(f"✅ Columns: {', '.join(df_real.columns)}")
print("=" * 100)

# Display file info
import os
file_size = os.path.getsize(csv_filename) / 1024  # Convert to KB
print(f"\n📊 File Size: {file_size:.2f} KB")
print(f"📅 Scraped Date: {df_real['Scraped_Date'].iloc[0]}")
print(f"⏰ Scraped Time: {df_real['Scraped_Time'].iloc[0]}")
print(f"\n✨ REAL BBC NEWS DATA SUCCESSFULLY SCRAPED AND SAVED!")
```
