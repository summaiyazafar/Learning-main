```python
# ============================================================
# 1. WHAT IS A VARIABLE?
# ============================================================

print("\n1. WHAT IS A VARIABLE?")
print("-" * 40)

# A variable stores data
name = "Ali"
age = 21
height = 5.9
is_student = True

print("Name:", name)
print("Age:", age)
print("Height:", height)
print("Is student:", is_student)
```

```python
# ============================================================
# 2. VARIABLE TYPES
# ============================================================

print("\n2. VARIABLE TYPES")
print("-" * 40)

integer_var = 10
float_var = 3.14
string_var = "Python"
boolean_var = False
complex_var = 2 + 3j

print(integer_var, type(integer_var))
print(float_var, type(float_var))
print(string_var, type(string_var))
print(boolean_var, type(boolean_var))
print(complex_var, type(complex_var))
```

```python
# ============================================================
# 3. DYNAMIC TYPING
# ============================================================

print("\n3. DYNAMIC TYPING")
print("-" * 40)

x = 10
print("x =", x, "| type:", type(x))

x = "Now I am a string"
print("x =", x, "| type:", type(x))

x = 3.14159
print("x =", x, "| type:", type(x))
```

```python
# ============================================================
# 5. MULTIPLE ASSIGNMENT
# ============================================================

print("\n5. MULTIPLE ASSIGNMENT")
print("-" * 40)

a, b, c = 1, 2, 3

print("a =", a)
print("b =", b)
print("c =", c)

# Same value to multiple variables
x = y = z = 100

print("x =", x)
print("y =", y)
print("z =", z)
```

```python
# ============================================================
# 6. VARIABLE SWAPPING
# ============================================================

print("\n6. VARIABLE SWAPPING")
print("-" * 40)

first = "Tea"
second = "Coffee"

print("Before swap:")
print("first =", first)
print("second =", second)

# Swapping
first, second = second, first

print("After swap:")
print("first =", first)
print("second =", second)
```

```python
# ============================================================
# 8. TYPE CASTING
# ============================================================

print("\n8. TYPE CASTING")
print("-" * 40)

num_str = "25"

print("Original:", num_str, type(num_str))

converted_num = int(num_str)

print("Converted:", converted_num, type(converted_num))

# More casting
print(float(10))
print(str(99))
print(bool(1))
print(bool(0))

# ============================================================
# 9. MEMORY ADDRESS (id)
# ============================================================

print("\n9. MEMORY ADDRESS")
print("-" * 40)

value1 = 100
value2 = 100

print("value1 id:", id(value1))
print("value2 id:", id(value2))

print("Same object in memory?", id(value1) == id(value2))

# ============================================================
# 10. MUTABLE VS IMMUTABLE VARIABLES
# ============================================================

print("\n10. MUTABLE VS IMMUTABLE")
print("-" * 40)

# Immutable example
text = "Hello"

print("Original text:", text)
print("Original id:", id(text))

text = text + " World"

print("Modified text:", text)
print("New id:", id(text))

print("\nStrings are immutable.")

# Mutable example
numbers = [1, 2, 3]

print("\nOriginal list:", numbers)
print("Original id:", id(numbers))

numbers.append(4)

print("Modified list:", numbers)
print("Same id after modification:", id(numbers))

print("Lists are mutable.")

# ============================================================
# 11. VARIABLE SCOPE
# ============================================================

print("\n11. VARIABLE SCOPE")
print("-" * 40)

global_var = "I am global"

def my_function():
    local_var = "I am local"
    print(global_var)
    print(local_var)

my_function()

# print(local_var)  # Would cause error

# ============================================================
# 12. GLOBAL KEYWORD
# ============================================================

print("\n12. GLOBAL KEYWORD")
print("-" * 40)

counter = 0

def increase():
    global counter
    counter += 1

increase()
increase()

print("Counter:", counter)

# ============================================================
# 13. CONSTANTS (CONVENTION)
# ============================================================

print("\n13. CONSTANTS")
print("-" * 40)

# Python does not have true constants
# Uppercase names are used by convention

PI = 3.14159
MAX_USERS = 100

print("PI =", PI)
print("MAX_USERS =", MAX_USERS)

# ============================================================
# 14. DELETING VARIABLES
# ============================================================

print("\n14. DELETING VARIABLES")
print("-" * 40)

temp_var = "Temporary"

print("Before delete:", temp_var)

del temp_var

# print(temp_var)  # Would cause error

print("Variable deleted successfully.")

# ============================================================
# 15. NONE TYPE
# ============================================================

print("\n15. NONE TYPE")
print("-" * 40)

data = None

print("data =", data)
print("Type:", type(data))

# ============================================================
# 16. VARIABLES INSIDE DATA STRUCTURES
# ============================================================

print("\n16. VARIABLES INSIDE DATA STRUCTURES")
print("-" * 40)

# List
fruits = ["apple", "banana", "orange"]

# Tuple
coordinates = (10, 20)

# Dictionary
student = {
    "name": "Ali",
    "age": 22
}

# Set
unique_numbers = {1, 2, 3}

print("List:", fruits)
print("Tuple:", coordinates)
print("Dictionary:", student)
print("Set:", unique_numbers)

# ============================================================
# 17. REFERENCE BEHAVIOR
# ============================================================

print("\n17. REFERENCE BEHAVIOR")
print("-" * 40)

list1 = [1, 2, 3]
list2 = list1

print("Before modification:")
print("list1 =", list1)
print("list2 =", list2)

list2.append(4)

print("\nAfter modifying list2:")
print("list1 =", list1)
print("list2 =", list2)

print("\nBoth variables reference same object.")

# ============================================================
# 18. COPYING VARIABLES
# ============================================================

print("\n18. COPYING VARIABLES")
print("-" * 40)

original = [1, 2, 3]

# Copy list
copied = original.copy()

copied.append(999)

print("Original:", original)
print("Copied:", copied)

# ============================================================
# 19. FORMATTED VARIABLES IN STRINGS
# ============================================================

print("\n19. FORMATTED STRINGS")
print("-" * 40)

language = "Python"
version = 3.12

message = f"I am learning {language} version {version}"

print(message)

# ============================================================
# 20. CHECKING VARIABLE TYPES
# ============================================================

print("\n20. TYPE CHECKING")
print("-" * 40)

score = 95

print(isinstance(score, int))
print(isinstance(score, str))

# ============================================================
# 21. PRACTICAL MINI PROJECT
# ============================================================

print("\n21. MINI PROJECT")
print("-" * 40)

student_name = "Sara"
student_marks = 88
passing_marks = 50

has_passed = student_marks >= passing_marks

print(f"Student: {student_name}")
print(f"Marks: {student_marks}")
print(f"Passed: {has_passed}")

# ============================================================
# 22. COMMON BEGINNER MISTAKES
# ============================================================

print("\n22. COMMON BEGINNER MISTAKES")
print("-" * 40)

# Mistake 1: Case sensitivity
name = "Ali"
Name = "Ahmed"

print(name)
print(Name)

# Mistake 2: Using reserved keywords
# class = "Math"  # Invalid

# Mistake 3: Forgetting type conversion
num = "5"

print("Without conversion:", num + num)
print("With conversion:", int(num) + int(num))

# ============================================================
# 23. ADVANCED: UNPACKING
# ============================================================

print("\n23. UNPACKING")
print("-" * 40)

person = ("Ali", 22, "Engineer")

name, age, profession = person

print(name)
print(age)
print(profession)

# ============================================================
# 24. ADVANCED: STAR UNPACKING
# ============================================================

print("\n24. STAR UNPACKING")
print("-" * 40)

numbers = [1, 2, 3, 4, 5]

first, *middle, last = numbers

print("First:", first)
print("Middle:", middle)
print("Last:", last)

# ============================================================
# 25. EVERYTHING TOGETHER
# ============================================================

print("\n25. FINAL COMBINED EXAMPLE")
print("-" * 40)

user_name = "Ayesha"
user_age = 20
courses = ["Python", "AI", "Data Science"]

print(f"""
User Information
----------------
Name: {user_name}
Age: {user_age}
Courses: {courses}
Total Courses: {len(courses)}
""")

# ============================================================
# FINAL SUMMARY
# ============================================================

print("=" * 60)
print("SUMMARY")
print("=" * 60)

summary = """
You learned:
✓ What variables are
✓ Variable types
✓ Dynamic typing
✓ Naming rules
✓ Multiple assignment
✓ Swapping variables
✓ Type casting
✓ Mutable vs immutable
✓ Variable scope
✓ Global variables
✓ Constants
✓ Deleting variables
✓ None type
✓ References and copying
✓ String formatting
✓ Unpacking
✓ Common mistakes
"""

print(summary)

print("=" * 60)
print("END OF PYTHON VARIABLES MASTERCLASS")
print("=" * 60)
```

```python
#17. In C++

#include <iostream> ---> Includes the input/output library. Gives access to std::cout for printing to the console.
#include <vector> ---> Includes the vector library. Gives access to std::vector, a dynamic array container

int main() { #The program execution starts in the main() function.
    std::cout << "\n17. REFERENCE BEHAVIOR\n"; #Prints a title. \n inserts a newline.
    std::cout << std::string(40, '-') << '\n'; #Creates a string containing 40 dashes (-). Prints it followed by a newline.

    std::vector<int> list1 = {1, 2, 3}; #Creates a vector named list1. Initializes it with the values 1, 2, and 3.
    std::vector<int>& list2 = list1;  // reference to list1 #Creates a reference named list2. The & here means "reference", not address. list2 does not create a new vector. list2 becomes another name for list1.

    std::cout << "Before modification:\n";

    std::cout << "list1 = ";
    for (int x : list1) std::cout << x << " "; #Range-based for loop. Iterates through every element in list1. Prints each value followed by a space.
    std::cout << '\n';

    std::cout << "list2 = ";
    for (int x : list2) std::cout << x << " ";
    std::cout << '\n';

    list2.push_back(4); #Adds the value 4 to the vector. Since list2 is a reference to list1, the actual vector list1 is modified.

    std::cout << "\nAfter modifying list2:\n";

    std::cout << "list1 = ";
    for (int x : list1) std::cout << x << " ";
    std::cout << '\n';

    std::cout << "list2 = ";
    for (int x : list2) std::cout << x << " ";
    std::cout << '\n';

    std::cout << "\nBoth variables reference the same object.\n";

    return 0;
}
```
