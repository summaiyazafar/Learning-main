```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.cluster import AgglomerativeClustering


# load the dataset(real dataset, not synthatic)
data= load_iris()
X= pd.DataFrame(data.data, columns=data.feature_names)

print("Dataset shape:", X.shape)
print("\nFirst 5 rows:\n", X.head())
```

```python
# Fit the Hierarchical (Agglomerative) clustering
hc= AgglomerativeClustering(
    n_clusters=3, # no. of clustering to extract 
    linkage='ward' # minimizes within-cluster variance
)
```

```python
# Fit the model and predict clusters
cluster_labels = hc.fit_predict(X)

# Attach cluster labels to dataset
X['Cluster'] = cluster_labels

print("\nCluster Count:")
print(X['Cluster'].value_counts())
```
